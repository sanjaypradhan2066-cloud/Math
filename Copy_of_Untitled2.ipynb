```python
def fib(n):
  a=0
  b=1
  i=0
  while i<n:
    print(a)
    c=a+b
    a=b
    b=c
    i=i+1
n=int(input("enter the number of terms"))
fib(n)
```

```python
x=int(input("enter base number"))
n=int(input("enter power of base number"))
def myPow(x,n):
  result=1
  i=0
  while i<n:
    result = result*x
    i=i+1
  return result

print(myPow(x,n))
```

```python
n=int(input("enter a number"))
def SumDigits(n):
  sum=0
  while n!=0:
    s=n%10
    sum+=s
    n=n//10

  return sum

print(SumDigits(n))
```

```python
def squaredSum(numbers):
    total = 0
    for num in numbers:
        total = total + (num * num)
    return total
print(squaredSum([2, 3, 4]))
```

```python
n = int(input("Enter a number: "))

def sumDigits(n):
    total = 0

    while n > 0:
        digit = n % 10
        total = total + digit
        n = n // 10

    return total

print(sumDigits(n))
```
